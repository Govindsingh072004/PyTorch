```python
import numpy as np
import pandas as pd
import seaborn as sns
import torch
import torch.nn as nn
import torch.nn.functional as F
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
##convert Outcome into Categorical Features for seaborn
```

```python
df["Outcome"]=np.where(df['Outcome']==1,"Diabetic","No Diabetic") ##optinal
```

```python
df.head()
```

```python
sns.pairplot(df,hue="Outcome")
```

```python
X=df.drop("Outcome",axis=1).values
y=df["Outcome"].values
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.4,random_state=0)
```

## start creating with PyTorch

```python
#create tensors
X_train=torch.FloatTensor(X_train)
X_test=torch.FloatTensor(X_test)
y_train=torch.Tensor(y_train)
y_test=torch.Tensor(y_test)
```

```python
X_train.shape
```

```python
#create Model with Pytorch
```

```python
class ANN_Model(nn.Module):
    def __init__(self,input_features=8,hidden1=20,hidden2=20,out_features=2):
        super().__init__()
        self.f_connected1=nn.Linear(input_features,hidden1)
        self.f_connected2=nn.Linear(hidden1,hidden2)
        self.out=nn.Linear(hidden2,out_features)
    def forward(self,x):
        x=F.relu(self.f_connected1(x))
        x=F.relu(self.f_connected2(x))
        x=self.out(x)
        return x
```

```python
torch.manual_seed(20)
model=ANN_Model()
```

```python
a=model.parameters
```

```python
##backPropogation --define the loss and optimizer
```

```python
loss_funcation=nn.CrossEntropyLoss()
optimizer=torch.optim.Adam(model.parameters(),lr=0.01)
```

```python
y_train=y_train.long()
```

```python
epochs=500
final_losses=[]
for i in range(epochs):
    i=i+1
    y_pred=model.forward(X_train)
    loss=loss_funcation(y_pred,y_train)
    final_losses.append(loss)
    if i%10==1:
        print("Epoch numbe: {} and the loss : {}".format(i,loss.item()))
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    
```

```python
# Detach tensors and convert to NumPy for plotting
final_losses_np = [loss.detach().numpy() for loss in final_losses]

# Plot the losses
plt.plot(range(len(final_losses_np)), final_losses_np)
plt.ylabel("Loss")
plt.xlabel("Epoch")
plt.title("Loss over Epochs")
plt.show()
```

```python
##prediction in X_test data
predictions=[]
with torch.no_grad():
    for i,data in enumerate(X_test):
        y_pred=model(data)
        predictions.append(y_pred.argmax().item())
        print(y_pred.argmax().item())
        
```

```python
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(y_test,predictions)
cm
```

```python
print(f"y_test length: {len(y_test)}")
print(f"predictions length: {len(predictions)}")
print(predictions)  # To check the actual content
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(cm,annot=True)
plt.xlabel("Actual Values")
plt.ylabel("Predicted Values")
```

```python
from sklearn.metrics import accuracy_score
score=accuracy_score(y_test,predictions)
```
